```python
# STEP 43 - Correlation Matrix

numeric_columns = [
    "Farm_Area_Hectares",
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Sunlight_Hours_Day",
    "Soil_pH",
    "Soil_Moisture_pct",
    "Nitrogen_kg_ha",
    "Phosphorus_kg_ha",
    "Potassium_kg_ha",
    "Fertilizer_kg_ha",
    "Pesticide_Litre_ha",
    "Seed_Quality_Score",
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Market_Price_INR_Tonne",
    "Total_Cost_INR",
    "Revenue_INR",
    "Profit_INR",
    "Water_Used_m3",
    "Water_Efficiency_t_per_1000m3",
    "Disease_Pest_Risk_pct"
]

correlation = df[numeric_columns].corr()

print(correlation)
```

```python
# STEP 44 - Correlation Heatmap

plt.figure(figsize=(16,12))

plt.imshow(correlation, aspect="auto")

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```

```python
# STEP 45 - Correlation with Yield

yield_correlation = correlation["Yield_Tonnes_Ha"].sort_values(ascending=False)

print(yield_correlation)
```

```python
# STEP 46 - Correlation with Profit

profit_correlation = correlation["Profit_INR"].sort_values(ascending=False)

print(profit_correlation)
```

```python
# STEP 47 - Season and Crop Yield

season_crop_yield = pd.pivot_table(
    df,
    values="Yield_Tonnes_Ha",
    index="Season",
    columns="Crop",
    aggfunc="mean"
)

print(season_crop_yield)

plt.figure(figsize=(12,6))

plt.imshow(season_crop_yield, aspect="auto")

plt.colorbar(label="Yield (Tonnes/Ha)")

plt.xticks(
    range(len(season_crop_yield.columns)),
    season_crop_yield.columns,
    rotation=45
)

plt.yticks(
    range(len(season_crop_yield.index)),
    season_crop_yield.index
)

plt.title("Average Yield by Season and Crop")

plt.tight_layout()
plt.show()
```

```python
# STEP 48 - Season and Crop Profit

season_crop_profit = pd.pivot_table(
    df,
    values="Profit_INR",
    index="Season",
    columns="Crop",
    aggfunc="mean"
)

print(season_crop_profit)

plt.figure(figsize=(12,6))

plt.imshow(season_crop_profit, aspect="auto")

plt.colorbar(label="Profit (INR)")

plt.xticks(
    range(len(season_crop_profit.columns)),
    season_crop_profit.columns,
    rotation=45
)

plt.yticks(
    range(len(season_crop_profit.index)),
    season_crop_profit.index
)

plt.title("Average Profit by Season and Crop")

plt.tight_layout()
plt.show()
```

```python
# STEP 49 - Final Season Summary

season_summary = df.groupby("Season").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "mean",
    "Total_Cost_INR": "mean",
    "Revenue_INR": "mean",
    "Profit_INR": "mean",
    "Water_Used_m3": "mean",
    "Water_Efficiency_t_per_1000m3": "mean",
    "Disease_Pest_Risk_pct": "mean"
}).round(2)

print(season_summary)
```

```python
# STEP 50 - Best Performing Season and Crop

best_season_yield = df.groupby("Season")["Yield_Tonnes_Ha"].mean().idxmax()

best_season_profit = df.groupby("Season")["Profit_INR"].mean().idxmax()

best_crop_yield = df.groupby("Crop")["Yield_Tonnes_Ha"].mean().idxmax()

best_crop_profit = df.groupby("Crop")["Profit_INR"].mean().idxmax()

print("Best Season by Yield:", best_season_yield)
print("Best Season by Profit:", best_season_profit)

print("Best Crop by Yield:", best_crop_yield)
print("Best Crop by Profit:", best_crop_profit)
```
